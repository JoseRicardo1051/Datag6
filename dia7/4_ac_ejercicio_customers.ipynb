# BUSCAMOS OUTLIERS

```python
from scipy.stats import zscore

z_scores = df.apply(zscore)

outliers = (z_scores.abs() > 3).sum()

outliers
```

```python
plt.figure(figsize=(12,6))

sns.boxplot(data=df[['Age','Education','Income']])
plt.show()
```

```python
df_cleaned = df[
    (z_scores['Age'].abs() <= 3) &
    (z_scores['Education'].abs() <= 3) &
    (z_scores['Income'].abs() <= 3)
].copy()
df_cleaned
```

# ELIMINAMOS COLUMNAS INNECESARIAS

```python
df_cleaned.drop(columns=['ID'],inplace=True)
df_cleaned.head(5)
```

# ESCALAMOS VALORES

```python
X = df_cleaned.copy()
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

##  Aplicar Agglomerative Clustering

```python
from sklearn.cluster import AgglomerativeClustering
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score

from scipy.cluster.hierarchy import dendrogram
```

```python
model = AgglomerativeClustering(
    n_clusters=14,
    linkage="ward"
)
```

```python
labels = model.fit_predict(X_scaled)
```

```python
df_cleaned["cluster"] = labels
```

```python
df_cleaned['cluster'].value_counts()
```

```python
print("Clusters encontrados:", np.unique(labels))
```

```python
sil = silhouette_score(X_scaled, labels)
print("Silhouette Score:", round(sil, 3))
```
