## PASO 2 - CREAMOS EL DATASET

```python
dataset = [
    ("I love this product, it's amazing!", "pos"),
    ("This movie was fantastic and enjoyable", "pos"),
    ("What a great experience!", "pos"),
    ("I hate this service, very disappointing", "neg"),
    ("This is the worst thing I have bought", "neg"),
    ("Completely useless and bad", "neg")
]
```

## PASO 3 - PROCESAMOS LA DATA

```python
nltk.download('stopwords')
```

```python
def extraer_caracteristicas(texto):
    stop_words = set(stopwords.words('english'))
    palabras = word_tokenize(texto.lower())
    palabras_limpias = [p for p in palabras if p.isalpha() and p not in stop_words]
    return dict([(palabra, True) for palabra in palabras_limpias])
```

```python
datos_entrenamiento = [(extraer_caracteristicas(texto), etiqueta) for texto, etiqueta in dataset]
```

## PASO 4 - ENTRENAMOS EL MODELO CON NLTK

```python
modelo = NaiveBayesClassifier.train(datos_entrenamiento)
```

## PASO 5- PROBAMOS EL MODELO

```python
texto_prueba = "I really love this amazing movie!"
resultado = modelo.classify(extraer_caracteristicas(texto_prueba))
print(f"Texto: {texto_prueba}")
print(f"Sentimiento detectado: {resultado}")
```
