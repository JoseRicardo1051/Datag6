# 🎧 Modelos para generación de Audio

---

## 📍 ¿Para qué sirven los modelos de generación de audio?

Los **modelos de generación de audio** permiten **crear, transformar o analizar sonidos mediante inteligencia artificial**.  
Estos modelos pueden generar desde música, efectos de sonido y voz sintética, hasta traducciones y transcripciones automáticas.

### 🎯 Usos comunes:
- 🎵 **Composición musical asistida por IA**  
- 🗣️ **Síntesis de voz (Text-to-Speech)**  
- 🧠 **Reconocimiento automático del habla (Speech-to-Text)**  
- 🎬 **Creación de efectos de sonido para videojuegos o películas**

---

# Cargar el modelo liviano

```python
from transformers import AutoProcessor, MusicgenForConditionalGeneration
import torch
import scipy.io.wavfile

# Cargar procesador y modelo
processor = AutoProcessor.from_pretrained("facebook/musicgen-small")
model = MusicgenForConditionalGeneration.from_pretrained("facebook/musicgen-small")

# Detectar GPU si existe
device = "cuda" if torch.cuda.is_available() else "cpu"
model.to(device)
```

# Definir el texto (prompt)

```python
prompt = "A calm lo-fi hip hop beat for studying with soft drums and piano"
```

```python
# Preparar entrada
inputs = processor(text=prompt, return_tensors="pt").to(device)
```

# Generar el audio

```python
audio_values = model.generate(**inputs, max_new_tokens=512)

# Guardar archivo de salida
sampling_rate = model.config.audio_encoder.sampling_rate
output_path = "/content/musicgen_output.wav"
scipy.io.wavfile.write(output_path, rate=sampling_rate, data=audio_values[0, 0].cpu().numpy())

print("✅ Audio generado correctamente en:", output_path)
```
