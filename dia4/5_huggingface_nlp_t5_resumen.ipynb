# 2. Importar librerías

```python
from transformers import T5Tokenizer, T5ForConditionalGeneration
import torch
```

# 3. Seleccionar dispositivo

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Dispositivo:", device)
```

# 4. Cargar modelo y tokenizador

```python
modelo_nombre = "t5-small"

tokenizer = T5Tokenizer.from_pretrained(modelo_nombre)
model = T5ForConditionalGeneration.from_pretrained(modelo_nombre)
model = model.to(device)
```

# 5. Función para resumir

```python
def resumir_texto(texto, max_input_length=512, max_output_length=120,
                  min_output_length=30, num_beams=4):
    """
    Resume un texto usando T5.
    """

    # T5 espera una instrucción tipo texto -> texto
    prompt = "summarize: " + texto.strip()

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        max_length=max_input_length,
        truncation=True
    )

    input_ids = inputs["input_ids"].to(device)
    attention_mask = inputs["attention_mask"].to(device)

    resumen_ids = model.generate(
        input_ids=input_ids,
        attention_mask=attention_mask,
        max_length=max_output_length,
        min_length=min_output_length,
        num_beams=num_beams,
        length_penalty=2.0,
        early_stopping=True,
        no_repeat_ngram_size=2
    )

    resumen = tokenizer.decode(resumen_ids[0], skip_special_tokens=True)
    return resumen
```

# 6 probamos el modelo

```python
texto = """
La inteligencia artificial ha transformado múltiples industrias en los últimos años.
En educación, por ejemplo, permite crear plataformas personalizadas que adaptan el contenido
al ritmo de aprendizaje de cada estudiante. En medicina, ayuda a detectar enfermedades
de manera más rápida y precisa a partir del análisis de imágenes y datos clínicos.
Sin embargo, también existen retos importantes relacionados con la ética, la privacidad
de los datos y el impacto en el empleo. Por ello, es fundamental desarrollar soluciones
de inteligencia artificial responsables, transparentes y centradas en las personas.
"""
resumen = resumir_texto(texto)

print("TEXTO ORIGINAL:\n")
print(texto)
print("\n" + "="*80 + "\n")
print("RESUMEN:\n")
print(resumen)
```
