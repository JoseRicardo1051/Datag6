# IMPORTAMOS LIBRERIAS

```python
import keras
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# CARGAMOS EL MODELO

```python
red = keras.models.load_model('/content/cnn_neumonia.keras')
red.summary()
```

# CAPTURAMOS PATH DE IMAGENES DE PRUEBA

```python
path_imagen_normal = '/content/NORMAL2-IM-0052-0001.jpeg'
path_imagen_neumonia = '/content/ryct.2020200034.fig5-day4.jpeg'
```

# PROBAMOS EL MODELO CON LAS IMAGENES DE PRUEBA

```python
from keras.preprocessing import image
path_imagen_prueba = path_imagen_neumonia
imagen_prueba = image.load_img(path_imagen_prueba,target_size=(64,64))

test_image = image.img_to_array(imagen_prueba)
test_image = np.expand_dims(test_image,axis=0)

result = red.predict(test_image)
print(f'predicción : {result}')
```
