```python
# Use a pipeline as a high-level helper
from transformers import pipeline

pipe = pipeline("text-classification", model="cesarcodigo/cesarcodigo-toxicidad-g6")
```

```python
texto = "this is the uggly face that i ever see in my live, please go away"
respuesta = pipe(texto)
print(respuesta)
```

```python
def analizar_texto(texto):
  resultado = pipe(texto)[0]
  label = resultado['label']
  score = resultado['score']
  if label == 'LABEL_1':
    return f'This comment is toxic (score: {score})'
  else:
    return f'This comment is not toxic (score: {score}'
```

```python
print(analizar_texto(texto))
```

```python
import gradio as gr

demo = gr.Interface(
    fn=analizar_texto,
    inputs=gr.Textbox(label='write a comment'),
    outputs=gr.Textbox(label='result')
)

demo.launch()
```
